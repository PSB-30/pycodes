```python
# Define dependent and independent variables
time = data['study_time']   # Dependent variable (1D array)
score = data[['test_score']]  # Independent variable (2D array)

# Create and fit the model
regr = LinearRegression()
regr.fit(score, time)

# Print regression coefficients
print("Slope (theta_1):", regr.coef_[0])
print("Intercept (theta_0):", regr.intercept_)

# R-squared value
print("R-squared:", regr.score(score, time))

# Make predictions
predicted_time = regr.predict(score)

# Plot the data and regression line
plt.scatter(score, time, color='blue', label='Actual data')
plt.plot(score, predicted_time, color='red', label='Regression line')

plt.title('Test Score vs Study Time')
plt.xlabel('Test Score')
plt.ylabel('Study Time')
plt.legend()
plt.grid(True)
plt.show()
```
