```python
import pandas
from pandas import DataFrame
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data=pandas.read_csv("budget_data.csv")
```

```python
data
```

```python
data.describe()
```

#Check for DataFrame so we can plot scatter plot

```python
X=DataFrame(data,columns=['budget'])
y=DataFrame(data,columns=['revenue'])
```

#Visualize

```python
plt.scatter(X,y)
plt.show()
```

```python
plt.title('Movie Costing')
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X,y,alpha=0.3)
plt.xlabel("Budget")
plt.ylabel("Revenue")
plt.ylim(0,300000000)
plt.xlim(0,450000000)
plt.show()
```

#Algorithm application

```python
regression = LinearRegression()

```

```python
regression.fit(X, y)
```

```python
regression.coef_ 
```

```python
#Intercept
regression.intercept_
```

```python
#Adding Liner Regression Line
plt.figure(figsize=(10,6))
plt.scatter(X,y,alpha=0.3)
plt.title("with Regression")
plt.plot(X,regression.predict(X),color='red',linewidth=3)
plt.xlabel("Budget")
plt.ylabel("Revenue")
plt.ylim(10,1000000000)
plt.xlim(200000000,450000000)
plt.show()
```
